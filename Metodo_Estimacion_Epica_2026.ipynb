```python
# Conteo de Datos Perdidos y % de Datos Perdidos
def get_numerical_summary(df):
    total = df.shape[0]
    missing_columns = [col for col in df.columns if (df[col].isnull().sum() > 0)]
    missing_percent = {}
    for col in missing_columns:
        null_count = df[col].isnull().sum()
        per = (null_count/total) * 100
        missing_percent[col] = per
        print("{} : {} ({}%)".format(col, null_count, round(per, 3)))
    return missing_percent
```

```python
# Reportar la estimacion de conteo y % de datos perdidos
missing_percent = get_numerical_summary(df_sprints)
```

```python
# Strip any leading/trailing whitespace from column names in df_sprints_new
df_sprints_new.columns = df_sprints_new.columns.str.strip()

selected_columns = [
    'Summary','Issue key', 'Issue id', 'Issue Type', 'Status', 'Project key', 'Project name',
    'Project type', 'Project lead', 'Project lead id', 'Description', 'Priority', 'Resolution',
    'Assignee', 'Reporter', 'Sprint', 'Status Category', 'Status Category Changed',
    'Custom field (Story Points)', 'Custom field (Criterios de aceptación)',
    'Custom field (Reglas de Negocio)', 'Custom field (Solución Técnica)', 'Created'
]

print("Columns in df_sprints_new after stripping whitespace:")
print(df_sprints_new.columns.tolist())

# Identify actually existing columns to be robust against missing ones
actual_selected_columns = [col for col in selected_columns if col in df_sprints_new.columns]

if not actual_selected_columns:
    raise KeyError("No valid columns from 'selected_columns' found in 'df_sprints_new' after cleaning column names.")
elif len(actual_selected_columns) < len(selected_columns):
    missing_cols = [col for col in selected_columns if col not in df_sprints_new.columns]
    print(f"Warning: The following requested columns were not found in df_sprints_new: {missing_cols}")

# Create the new DataFrame with only the actual selected columns
df_selected_variables = df_sprints_new[actual_selected_columns].copy()

print("New DataFrame 'df_selected_variables' created with the specified columns.")
print("Columns in the new DataFrame:")
print(df_selected_variables.columns.tolist())
print("First 5 rows of the new DataFrame:")
print(df_selected_variables.head())
```

```python
import re

# Assuming the year is 2025 based on the dataset name '30122025-historias-bugs.csv'
year = '2025'

# Function to convert Spanish month abbreviations to numbers (case-insensitive)
def get_month_number(month_abbr):
    month_map = {
        'ene': '01', 'feb': '02', 'mar': '03', 'abr': '04', 'may': '05', 'jun': '06',
        'jul': '07', 'ago': '08', 'sep': '09', 'oct': '10', 'nov': '11', 'dic': '12'
    }
    return month_map.get(month_abbr.lower(), None) # Convert to lowercase for lookup

# Function to extract start and end dates from the Sprint string
def extract_sprint_dates(sprint_str):
    if pd.isna(sprint_str):
        return pd.NaT, pd.NaT

    # Regex to find two dates in the format 'DD Mon a DD Mon' or 'DDMon a DDMon'
    # Using re.IGNORECASE to match 'Ene', 'ene', 'NOV', 'nov' etc.
    # Using \s? to make the space character optional between day and month
    match = re.search(r'(\d{1,2})\s?(Ene|Feb|Mar|Abr|May|Jun|Jul|Ago|Sep|Oct|Nov|Dic)\s(?:a|al)\s(\d{1,2})\s?(Ene|Feb|Mar|Abr|May|Jun|Jul|Ago|Sep|Oct|Nov|Dic)', sprint_str, re.IGNORECASE)
    if match:
        start_day = match.group(1)
        start_month_abbr = match.group(2)
        end_day = match.group(3)
        end_month_abbr = match.group(4)

        start_month_num = get_month_number(start_month_abbr)
        end_month_num = get_month_number(end_month_abbr)

        if start_month_num and end_month_num:
            try:
                start_date_str = f'{start_day}/{start_month_num}/{year}'
                end_date_str = f'{end_day}/{end_month_num}/{year}'
                start_date = pd.to_datetime(start_date_str, format='%d/%m/%Y')
                end_date = pd.to_datetime(end_date_str, format='%d/%m/%Y')
                return start_date, end_date
            except ValueError:
                return pd.NaT, pd.NaT
    return pd.NaT, pd.NaT

# Apply the function to the 'Sprint' column to create new date columns
df_selected_variables[['Fecha_inicio_Sprint', 'Fecha_fin_Sprint']] = df_selected_variables['Sprint'].apply(lambda x: pd.Series(extract_sprint_dates(x)))

print("New variables 'Fecha_inicio_Sprint' and 'Fecha_fin_Sprint' added to df_selected_variables.")
print("First 5 rows with 'Sprint', 'Fecha_inicio_Sprint', and 'Fecha_fin_Sprint' variables:")
print(df_selected_variables[['Sprint', 'Fecha_inicio_Sprint', 'Fecha_fin_Sprint']].head())
```

```python
numeric_cols = df_selected_variables.select_dtypes(include=['number']).columns.tolist()
categorical_cols = df_selected_variables.select_dtypes(exclude=['number', 'datetime']).columns.tolist()

print("Columnas numéricas:")
print(numeric_cols)
print("\nColumnas categóricas:")
print(categorical_cols)
```

```python
for col in numeric_cols:
    if df_selected_variables[col].isnull().any():
        median_val = df_selected_variables[col].median()
        df_selected_variables[col] = df_selected_variables[col].fillna(median_val) # Remove inplace=True
        print(f"Imputed missing values in numeric column '{col}' with median: {median_val}")

# Keep track of columns already processed to avoid redundant imputation for duplicated names in categorical_cols list
processed_categorical_cols = set()
for col in categorical_cols:
    if col in processed_categorical_cols:
        continue

    # Access the column(s) by name. If it's a duplicated column name (like 'Description'), this returns a DataFrame.
    column_data = df_selected_variables[col]

    if isinstance(column_data, pd.DataFrame):
        # This handles the case where 'col' selects multiple columns (e.g., 'Description' is duplicated)
        # Check if any of these duplicated columns have any nulls, converting the resulting Series to a single boolean
        if column_data.isnull().any().any(): # Fixes ValueError: ambiguous Series in if condition
            # Get the mode from the first instance of the duplicated column to use for imputation
            mode_val = column_data.iloc[:, 0].mode()[0]
            # Impute all columns that match the name 'col' with this mode
            for sub_col_name in column_data.columns:
                df_selected_variables[sub_col_name] = df_selected_variables[sub_col_name].fillna(mode_val)
                print(f"Imputed missing values in categorical column '{sub_col_name}' (duplicated via '{col}') with mode: {mode_val}")
    else:
        # This handles the case where 'col' selects a single Series (a unique column name)
        if column_data.isnull().any():
            mode_val = column_data.mode()[0]
            df_selected_variables[col] = column_data.fillna(mode_val) # Remove inplace=True
            print(f"Imputed missing values in categorical column '{col}' with mode: {mode_val}")

    processed_categorical_cols.add(col) # Mark as processed

# Verify no missing values remain
missing_after_imputation = df_selected_variables.isnull().sum()
missing_after_imputation = missing_after_imputation[missing_after_imputation > 0]

if missing_after_imputation.empty:
    print("\nAll missing values have been successfully imputed.")
else:
    print("\nMissing values still exist in the following columns:")
    print(missing_after_imputation)
```

```python
import numpy as np

# Get unique assignees
unique_assignees = df_selected_variables['Assignee'].unique()

# Generate a random hourly wage for each unique assignee between 200 and 400
np.random.seed(42) # for reproducibility
assignee_hourly_wage = {
    assignee: np.random.randint(200, 401) for assignee in unique_assignees
}

# Map the hourly wages to the DataFrame based on the 'Assignee' column
df_selected_variables['hourly_wage'] = df_selected_variables['Assignee'].map(assignee_hourly_wage)

# Calculate 'Costo_por_dia' assuming an 8-hour workday
df_selected_variables['Costo_por_dia'] = df_selected_variables['hourly_wage'] * 8

print("New variable 'Costo_por_dia' added to df_selected_variables.")
print("First 5 rows with 'hourly_wage' and 'Costo_por_dia' variables:")
display(df_selected_variables[['Assignee', 'hourly_wage', 'Costo_por_dia']].head())
```

```python
# Calculate Score_esf
df_selected_variables['Score_esf'] = df_selected_variables['Costo_por_dia'] * df_selected_variables['tiempo'] * df_selected_variables['Custom field (Story Points)']

# Normalize Score_esf to a 1-10 scale
min_score_esf = df_selected_variables['Score_esf'].min()
max_score_esf = df_selected_variables['Score_esf'].max()

# Apply min-max scaling to convert Score_esf to Esfuerzo between 1 and 10
# The formula for min-max scaling to a new range [a, b] is:
# new_value = (original_value - min_original) / (max_original - min_original) * (b - a) + a
a = 1  # new minimum
b = 10 # new maximum

# Handle cases where max_score_esf == min_score_esf to avoid division by zero
if max_score_esf == min_score_esf:
    df_selected_variables['Esfuerzo'] = (a + b) / 2  # Assign midpoint if all scores are the same
else:
    df_selected_variables['Esfuerzo'] = ((df_selected_variables['Score_esf'] - min_score_esf) / (max_score_esf - min_score_esf)) * (b - a) + a

print("New variable 'Score_esf' and 'Esfuerzo' added to df_selected_variables.")
print("First 5 rows with 'Costo_por_dia', 'tiempo', 'Custom field (Story Points)', 'Score_esf', and 'Esfuerzo' variables:")
print(df_selected_variables[['Costo_por_dia', 'tiempo', 'Custom field (Story Points)', 'Score_esf', 'Esfuerzo']].head())
```

```python
df_selected_variables['Duracion_Sprint'] = df_selected_variables['tiempo']
df_selected_variables['Velocidad'] = df_selected_variables['Custom field (Story Points)'] / df_selected_variables['Duracion_Sprint']
df_selected_variables['Costo_por_Puntos_Historia'] = df_selected_variables['Costo_por_dia'] / df_selected_variables['Custom field (Story Points)']
df_selected_variables['Esfuerzo_por_Puntos_Historia'] = df_selected_variables['Esfuerzo'] / df_selected_variables['Custom field (Story Points)']

print("\nActualizar la base de datos:")
df_selected_variables.head()
```

```python
import pandas as pd
import numpy as np

TEXT_COLS = [
    "Summary",
    "Description",
    "Custom field (Criterios de aceptación)",
    "Custom field (Reglas de Negocio)",
    "Custom field (Solución Técnica)",
]

def build_text_corpus(df_done: pd.DataFrame,
                      points_col: str = "story_points",
                      key_col: str = "issue_key",
                      text_cols = TEXT_COLS,
                      min_chars: int = 40) -> pd.DataFrame:
    df = df_done.copy()

    # Validaciones
    missing = [c for c in text_cols if c not in df.columns]
    if missing:
        raise ValueError(f"Faltan columnas en df_done: {missing}")

    if points_col not in df.columns:
        raise ValueError(f"No encontré la columna de puntos '{points_col}' en df_done")

    if key_col not in df.columns:
        raise ValueError(f"No encontré la columna key '{key_col}' en df_done")

    # Normalizar puntos
    df["points"] = pd.to_numeric(
        df[points_col], errors="coerce"
    ).fillna(0.0)

    # Construir texto
    def join_text(row):
        parts = []
        for c in text_cols:
            v_raw = row.get(c, "") # Get the raw value, which could be a Series for duplicated cols

            # If v_raw is a Series (meaning a duplicated column name was accessed)
            if isinstance(v_raw, pd.Series):
                # Process each item in the Series
                for item in v_raw.dropna(): # Only consider non-null items
                    if str(item).strip():
                        parts.append(f"{c}: {str(item).strip()}")
            else:
                # Process as a single scalar value
                if pd.notna(v_raw) and str(v_raw).strip():
                    parts.append(f"{c}: {str(v_raw).strip()}")
        return "\n".join(parts)

    df["text"] = df.apply(join_text, axis=1)

    # Filtro: texto suficientemente descriptivo y puntos > 0 (opcional pero recomendado)
    corpus = df[(df["text"].str.len() >= min_chars) & (df["points"] > 0)].copy()

    corpus = corpus[[key_col, "text", "points"]].rename(columns={key_col: "issue_key"}).reset_index(drop=True)

    meta = {
        "rows_in": int(len(df)),
        "rows_out": int(len(corpus)),
        "pct_kept": float(len(corpus) / max(len(df), 1)),
        "points_sum": float(corpus["points"].sum()),
        "points_mean": float(corpus["points"].mean()) if len(corpus) else 0.0,
        "points_unique": int(corpus["points"].nunique()) if len(corpus) else 0,
        "min_chars": min_chars
    }
    return corpus, meta
```

## Preparar una funcion para descomponer Epicas

```python
from openai import AzureOpenAI
import json
import re

def decompose_epic_to_stories(
    epic_brief: str,
    n_min: int = 8,
    n_max: int = 20
) -> dict:
    """
    Usa Chat Completions para descomponer una épica en historias.
    Devuelve SIEMPRE un dict de Python (no string).
    """

    prompt = f"""
Eres un analista ágil. Descompón la siguiente ÉPICA en historias de usuario LIBERABLES a productivo. Debes desmenuzar la épica en el número suficiente de historias de usuario lo más detallado posible para
evitar una subestimación y contemplar todo el esfuerzo.

Reglas:
- Entrega entre {n_min} y {n_max} historias.
- Cada historia debe incluir:
  - id
  - titulo
  - descripcion
  - criterios_aceptacion (lista)
  - supuestos (lista)
  - tags (lista)
- NO asignes story points.
- Responde SOLO en JSON con la siguiente forma:

{{
  "historias_usuario": [ ... ]
}}

ÉPICA:
{epic_brief}
"""

    resp = client.chat.completions.create(
        model="gpt5.1-demo-master-ia-model",
        temperature=0.2,
        messages=[
            {"role": "system", "content": "Responde únicamente JSON válido. No agregues texto extra."},
            {"role": "user", "content": prompt}
        ]
    )

    raw = resp.choices[0].message.content

    # ---- PARSER ROBUSTO ----
    # 1) Intento directo
    try:
        return json.loads(raw)
    except json.JSONDecodeError:
        pass

    # 2) Extraer bloque JSON si el modelo agregó texto extra
    match = re.search(r"\{.*\}", raw, re.DOTALL)
    if not match:
        raise ValueError("❌ No se encontró JSON en la respuesta del modelo")

    return json.loads(match.group(0))
```
